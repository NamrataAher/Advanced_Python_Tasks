```python
import pandas as pd

data = {
    "Match_No": ["IPL2367", "IPL2367", "IPL2367"],
    "Innings": [1, 1, 1],
    "Team": ["Delhi Capitals"] * 3,
    "Player_Name": ["Rilee Russouw", "Phil Salt", "Yash Dhull"],
    "Ballcount": ["0.1", "0.2", "0.3"],
    "Position": ["Unknown", "Unknown", "Unknown"],
    "Short_Description": [
        "Sample fielding event 1",
        "Sample fielding event 2",
        "Sample fielding event 3"
    ],
    "Pick": ["Not recorded"] * 3,
    "Throw": ["Not recorded"] * 3,
    "Runs": [0, 0, 0],
    "Overcount": [0, 0, 0],
    "Venue": ["Arun Jaitley Stadium"] * 3
}

df = pd.DataFrame(data)

print(df)
```

```python
df.to_csv("Fielding_Data.csv", index=False)

print("Dataset saved successfully!")
```

```python
# Sample performance data
performance = {
    "Player_Name": ["Rilee Russouw", "Phil Salt", "Yash Dhull"],
    "CP": [2, 1, 3],
    "GT": [1, 2, 1],
    "C": [1, 0, 2],
    "DC": [0, 1, 0],
    "ST": [0, 0, 0],
    "RO": [0, 1, 0],
    "MRO": [0, 0, 1],
    "DH": [1, 0, 0],
    "RS": [2, -1, 3]
}

scores = pd.DataFrame(performance)

# Calculate Performance Score
scores["PS"] = (
    scores["CP"] * 1
    + scores["GT"] * 1
    + scores["C"] * 3
    + scores["DC"] * -3
    + scores["ST"] * 3
    + scores["RO"] * 3
    + scores["MRO"] * -2
    + scores["DH"] * 2
    + scores["RS"]
)

print(scores[["Player_Name", "PS"]])
```

```python
plt.figure(figsize=(8, 5))

plt.bar(scores["Player_Name"], scores["PS"])

plt.title("Cricket Fielding Performance Analysis")
plt.xlabel("Player Name")
plt.ylabel("Performance Score")

plt.tight_layout()
plt.show()
```

```python
best_player = scores.loc[scores["PS"].idxmax(), "Player_Name"]
best_score = scores["PS"].max()

print("CRICKET FIELDING ANALYSIS")
print("-------------------------")
print("Players analyzed:", len(scores))
print("Highest Performance Score:", best_player)
print("Score:", best_score)

print("\nCONCLUSION")
print(f"{best_player} has the highest score in this sample dataset.")
print("More verified fielding data is needed for a reliable match analysis.")
```

```python
scores.to_csv("Fielding_Performance_Scores.csv", index=False)

print("Performance scores saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

performance = {
    "Player_Name": [
        "Rilee Russouw", "Phil Salt", "Yash Dhull",
        "Axer Patel", "Lalit Yadav", "Aman Khan",
        "Kuldeep Yadav"
    ],
    "CP":  [2, 1, 3, 2, 1, 4, 3],
    "GT":  [1, 2, 1, 3, 2, 1, 0],
    "C":   [1, 0, 2, 1, 1, 0, 1],
    "DC":  [0, 1, 0, 0, 0, 0, 1],
    "ST":  [0, 0, 0, 1, 0, 0, 0],
    "RO":  [0, 1, 0, 0, 0, 1, 0],
    "MRO": [0, 0, 1, 0, 0, 0, 0],
    "DH":  [1, 0, 0, 0, 1, 0, 1],
    "RS":  [2, -1, 3, 0, -2, 1, 4]
}

scores = pd.DataFrame(performance)

scores["PS"] = (
    scores["CP"] + scores["GT"]
    + scores["C"] * 3
    - scores["DC"] * 3
    + scores["ST"] * 3
    + scores["RO"] * 3
    - scores["MRO"] * 2
    + scores["DH"] * 2
    + scores["RS"]
)

print(scores.to_string(index=False))

scores.to_csv("Fielding_Performance_Scores.csv", index=False)
print("\nAll 7 players' scores saved successfully!")
```

```python
plt.figure(figsize=(10, 6))

plt.bar(scores["Player_Name"], scores["PS"])

plt.title("IPL Cricket Fielding Performance Analysis")
plt.xlabel("Player Name")
plt.ylabel("Performance Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("CRICKET FIELDING PERFORMANCE ANALYSIS")
print("=" * 40)

print("\n1. Highest Performance Score:")
print(scores.loc[scores["PS"].idxmax(), "Player_Name"],
      "-", scores["PS"].max(), "points")

print("\n2. Lowest Performance Score:")
print(scores.loc[scores["PS"].idxmin(), "Player_Name"],
      "-", scores["PS"].min(), "points")

print("\n3. Average Performance Score:")
print(round(scores["PS"].mean(), 2))

print("\n4. Players Ranked by Performance:")
print(
    scores[["Player_Name", "PS"]]
    .sort_values("PS", ascending=False)
    .to_string(index=False)
)
```

## Conclusion

This project analyzes the fielding performance of seven players using the sample performance matrix provided in the task.

Yash Dhull and Axer Patel achieved the highest performance score of 11 points. Phil Salt recorded the lowest score of 2 points. The average performance score was 8.29 points.

The analysis compares clean picks, good throws, catches, dropped catches, stumpings, run outs, missed run outs, direct hits, and runs saved. The bar graph helps visualize the differences in player performance.

This scoring method can help identify fielding contributions and areas for improvement. However, the results are based on the sample performance matrix and should not be treated as verified match statistics.
