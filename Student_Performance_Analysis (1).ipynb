```python

import pandas as pd

df = pd.read_csv(r"C:\Users\Namrata Aher\Downloads\Student_Performance_Data.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
df.describe()
```

```python

print(df.columns.tolist())
```

```python

average_scores = df[
    ["Assignment_Score", "Exam_Score"]
].mean()

average_scores.plot(
    kind="bar",
    title="Average Assignment Score vs Exam Score",
    xlabel="Assessment Type",
    ylabel="Average Score"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours_Per_Week"],
    df["Exam_Score"]
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance_Percent"],
    df["Exam_Score"]
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python

correlation = df[
    [
        "Study_Hours_Per_Week",
        "Attendance_Percent",
        "Assignment_Score",
        "Exam_Score"
    ]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Student Performance Factors")
plt.tight_layout()
plt.show()
```

```python

print("Total Students:", len(df))

print("\nAverage Study Hours:")
print(round(df["Study_Hours_Per_Week"].mean(), 2))

print("\nAverage Attendance:")
print(round(df["Attendance_Percent"].mean(), 2), "%")

print("\nAverage Assignment Score:")
print(round(df["Assignment_Score"].mean(), 2))

print("\nAverage Exam Score:")
print(round(df["Exam_Score"].mean(), 2))

print("\nHighest Exam Score:")
print(df["Exam_Score"].max())

print("\nLowest Exam Score:")
print(df["Exam_Score"].min())
```
